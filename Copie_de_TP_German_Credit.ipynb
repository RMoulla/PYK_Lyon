# **Travaux pratiques : analyse des données de crédit**

## Objectif du TP

Ce TP a pour objectif d’explorer et de nettoyer des données de crédit, d’étudier les facteurs associés au montant emprunté à l’aide de statistiques descriptives et de visualisations, puis de modéliser ce montant par régression linéaire.

## Description des variables

- **Age** : âge de la personne, en années.
- **Sex** : sexe de la personne (`male` ou `female`).
- **Job** : catégorie professionnelle, codée de 0 à 3 selon le niveau de qualification.
- **Housing** : situation de logement (`own`, `rent` ou `free`).
- **Saving accounts** : niveau des fonds disponibles sur le compte d’épargne (`little`, `moderate`, `quite rich`, `rich` ou information manquante).
- **Checking account** : niveau des fonds disponibles sur le compte courant (`little`, `moderate`, `rich` ou information manquante).
- **Credit amount** : montant du crédit.
- **Duration** : durée du crédit, en mois.
- **Purpose** : motif du crédit (véhicule, équipement, études, activité professionnelle, etc.).

## Modélisation

```python
df.head()
```

```python
# Transformer les variables catégorielles avec get_dummies

dataset = pd.get_dummies(df, columns=["Sex", "Housing", "Saving accounts", "Checking account", "Purpose"], drop_first=True).astype(int)
dataset.head()
```

```python
# Entrainer une régression linéaire avec statsmodels

import statsmodels.api as sm
from sklearn.model_selection import train_test_split

X = dataset.drop(columns=["Credit amount"])
y = dataset["Credit amount"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = sm.OLS(y_train, sm.add_constant(X_train)).fit()
print(model.summary())
```

```python
# Prédire y et calculer une rmse
from sklearn.metrics import root_mean_squared_error


y_pred = model.predict(sm.add_constant(X_test))

rmse = root_mean_squared_error(y_test, y_pred)

print(f"La RMSE est de : {rmse: .2f}")
```
