## Les conditions

```python
# si a est supérieru à 1 afficher "a est supérieur à 1"

a = 2
if a > 1:
  print("a est supérieur à 1")
```

```python
# si a est supérieur à 3 afficher "a est supérieur à 3"
# sinon afficher "a est inférieur ou égal à 3"

if a > 3:
  print("a est supérieur à 3")
else:
  print("a est inférieur ou égal à 3")
```

```python
# si a est égal à 2 afficher "a est égal à 2"

if a == 2:
  print("a est égal à 2")
```

```python
# condition complexe :
# si a n'est pas égal à 1
# et si a est inférieur à 3
# afficher a est inérieur à 3

if a != 1 and a < 3:
  print("a est inférieur à 3")
```

```python
# si a est différent de 1
# ou si a est inférieur à 0
# afficher "Bonjour"

if a != 1 or a < 0:
  print("Bonjour")
```

```python
# si a est égal à 1 afficher "a est égal à 1"
# sinon si a est inférieur à 1 afficher "afficher a est inférieur à 1"
# sinon si a est supérieur à 1 afficher "afficher a est supérieur à 1"

if a == 1:
  print("a est égal à 1")

elif a < 1:
  print("a est inférieur à 1")

else:
  print("a est supérieur à 1")
```

```python
# condition avec une chaine de caractère

char = "Bonjour"
if char:
  print("La chaine de caractère n'est pas vide")
```

```python
# Piège de la condition if sur les valeurs nulles

x = 0

if x:
  print("x a une valeur")
else:
  print("x n'a pas de valeur")
```

### Boucles itératives `for`

```python
# Boucle for qui calcule les carrés des nombres de ma liste

ma_liste = [1,2,3,4,5]

for iterateur in ma_liste:
  print(iterateur**2)
```

```python
# Boucle for avec range

for i in range(1,5):
  print(i)
```

### Les listes

```python
# Définir une liste

ma_liste = [10, 20, 40, 20, 5]
print(ma_liste)
```

```python
# Accéder au premier élément de la liste

ma_liste[0]
```

```python
# Accéder au dernier élément de la liste

ma_liste[-1]
```

```python
# Accéder à plusieurs éléments en focntion de l'index

ma_liste[1:3]
```

```python
# Modifier un élément d'une liste

ma_liste[0] = 15
print(ma_liste)
```

```python
# Insérer un élément dans une liste

ma_liste.insert(2, 35)
print(ma_liste)
```

```python
# Ajouter un élément à la dernière position

ma_liste.append(20)
print(ma_liste)
```

```python
# Supprimer un élément d'une liste avec pop

ma_liste.pop(2)
print(ma_liste)
```

```python
# Supprimer l'élément 20

ma_liste.remove(20)
print(ma_liste)
```

```python
# Insérer un élément dans une liste

ma_liste.insert(3, 20)
print(ma_liste)
```
