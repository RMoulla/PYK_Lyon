# **Travaux pratiques : analyse et prédiction du churn**

## Objectif du TP

Analyser les facteurs associés au départ des clients et construire un modèle de régression logistique capable de prédire le churn à partir de leurs caractéristiques.

## Description des variables

- **Names** : nom du client.
- **Age** : âge du client, en années.
- **Total_Purchase** : montant total des achats réalisés par le client.
- **Account_Manager** : indicateur de suivi par un gestionnaire de compte (`1` pour oui, `0` pour non).
- **Years** : ancienneté du client, en années.
- **Num_Sites** : nombre de sites associés au client.
- **Onboard_date** : date d'entrée du client.
- **Location** : adresse ou localisation du client.
- **Company** : entreprise du client.
- **Churn** : indicateur de départ du client (`1` pour un départ, `0` pour un client conservé).

```python
# Lire le fichier csv

import pandas as pd

df = pd.read_csv('customer_churn.csv')
df.head()
```

## Nettoyage des données

```python
# Supprimer les colonnes Names, Company, Location, Onboard_date

df = df.drop(['Names', 'Company', 'Location', 'Onboard_date'], axis=1)
df.head()
```

## Modélisation

```python
# Isoler les X et y

X = df.drop('Churn', axis=1)
y = df['Churn']

# Partitionner les données en train et test

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)
```

```python
# Entrainer un modèle de régression logistique avec statsmodels

import statsmodels.api as sm

X_train = sm.add_constant(X_train)
log_model = sm.Logit(y_train, X_train).fit()
print(log_model.summary())
```

```python
# Prédire sur l'échantillon de test

X_test = sm.add_constant(X_test)
y_pred = log_model.predict(X_test)


# Transformer les probabilités prédites en variable binaire (0 et 1)

y_pred_binary = [1 if prob > 0.25 else 0 for prob in y_pred]

# Calculer l'accuracy, la précision, le rappel et le F1 score
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_pred_binary)
precision = precision_score(y_test, y_pred_binary)
recall = recall_score(y_test, y_pred_binary)
f1 = f1_score(y_test, y_pred_binary)

print(f"Accuracy: {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1 Score: {f1:.2f}")
```
